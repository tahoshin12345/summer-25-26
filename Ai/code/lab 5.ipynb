```python
def createMatrix(n):
    m = []
    for i in range(n):
        row = []
        for j in range(n):
            row.append(0)
        m.append(row)
    return m

def printGraph(G):
    for row in G:
        print(row)

f = open("graph.txt", "r")
v_e = f.readline().split()
n = int(v_e[0])
m = int(v_e[1])

print(f"Vertices: {n}, Edges: {m}")


g = createMatrix(n)

for i in range(m):
    u_v = f.readline().split()
    u = int(u_v[0])
    v = int(u_v[1])
    g[u][v] = 1

f.close()

print("--- Adjacency Matrix ---")
printGraph(g)


def dfs(s, G):
    stack = []
    stack.append(s)
    explore_set = set()
    
    print("\nDFS Traversal Order:")
    while len(stack) != 0:
        node = stack.pop()
        
        if node not in explore_set:
            print(node, end=" ")
            explore_set.add(node)
            
            
            for i in range(len(G)):
                if G[node][i] == 1:
                    if i not in explore_set and i not in stack:
                        stack.append(i)


dfs(0, g)
```
