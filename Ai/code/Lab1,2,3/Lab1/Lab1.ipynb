```python
# Arithemetic Operator

a=10
b=20
print(a+b)
print(a-b)
print(a*b)
print(a/b) # floating division
print(a//b) # integer division
print(a**2) # exponent
print(a%b)
```

```python
# Relational Operator

a=10
b=20
print(a>b)
print(a==b)
print(a<=b)
print(a!=b)
```

```python
# Logical Operator

a=20
b=10
c=1

print((a>b) and (b<c))
print((a>b) or (b<c))
print((a>b) & (b<c))
print((a>b) | (b<c))
print(not (a>b))
```
