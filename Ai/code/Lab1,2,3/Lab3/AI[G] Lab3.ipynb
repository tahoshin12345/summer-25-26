```python
def changeList(L):
    for i in range(len(L)):
        L[i]=L[i]+2
    return L

def left_Shift(L):
    for i in range(len(L)-1):
        L[i]=L[i+1]
    L[-1]=0
    return L

def left_Rotate(L):
    v=L[0]
    for i in range(len(L)-1):  
        L[i]=L[i+1]
    L[-1]=v
    return L
    

List=[10,20,30,40,50]
new_List=changeList(List)
print(new_List)
new_List=left_Shift(List)
print(new_List)
new_List=left_Rotate(List)
print(new_List)
```

```python
def rotate(L,k):
    for j in range(k):
        v=L[0]
        for i in range(len(L)-1):
            L[i]=L[i+1]
        L[-1]=v
    return L

List=[10,20,30,40,50]
new_List=rotate(List,2)
print(new_List)
```

```python
import math
#θ = cos-1 [ (a · b) / (|a| |b|) ]
def angle(a,b):
    return math.acos(dot_product(a,b)/(magnitute(a)*magnitute(b)))

def dot_product(v1,v2):
    sum=0
    for i in range(len(v1)):
        sum=sum+v1[i]*v2[i]
    return sum
    
def magnitute(v):
    sum=0
    for i in range(len(v)):
        sum=sum+v[i]**2
    return math.sqrt(sum)

#|a × b| = |a| |b| sin θ
def cross_product(a,b):
    return magnitute(a)*magnitute(b)*math.sin(angle(a,b))

a=[10,20,30]
b=[1,20,3]
print(cross_product(a,b))
```
