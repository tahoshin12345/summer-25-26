# List of List
### Matrix

```python
r1=[10,20,30]
r2=[40,50,60]
r3=[70,80,90]
r4=[10,10,90]
m=[]
m.append(r1)
m.append(r2)
m.append(r3)
m.append(r4)
print(m)
print(len(m)) # row
print(len(m[0])) # column


for i in range(len(m)):
    for j in range(len(m[0])):
        print(m[i][j],end=" ")
    print()
```

```python
row=5
column=10
m=[]
for i in range(row):
    r=[]
    for j in range(column):
        r.append(0)
    m.append(r)

for i in range(len(m)):
    for j in range(len(m[0])):
        print(m[i][j],end=" ")
    print()
```
